```python
import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
data = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/letter-recognition/letter-recognition.data", header=None)
```

```python
X = data.iloc[:,1:].values
y = data.iloc[:,0].astype('category').cat.codes
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y)
```

```python
model = Sequential ([
    Dense(32,activation = 'relu',input_shape=(16,)),
    Dense(26,activation = 'softmax')
])
```

```python
model.compile(optimizer='adam', loss = 'sparse_categorical_crossentropy', metrics = ['accuracy'])
```

```python
model.fit(X_train,y_train,epochs = 5)
```

```python
print("The accuracy is",model.evaluate(X_test,y_test)[1])
```
